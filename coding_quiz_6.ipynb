```python
import pandas as pd
import numpy as np
from sklearn.neighbors import NearestNeighbors

df = pd.read_csv("homework_6.1.csv")
treated = df[df["X"] == 1].reset_index(drop=True)
untreated = df[df["X"] == 0].reset_index(drop=True)

def nearest_y(source, target):
    """For each row in source, return Y of the nearest row in target (matched on Z)."""
    nn = NearestNeighbors(n_neighbors=1).fit(target[["Z"]])
    _, idx = nn.kneighbors(source[["Z"]])
    return target["Y"].values[idx.ravel()]

# Treated items: actual Y (treated) minus counterfactual Y (nearest untreated)
effects_treated = treated["Y"].values - nearest_y(treated, untreated)

# Untreated items: counterfactual Y (nearest treated) minus actual Y (untreated)
effects_untreated = nearest_y(untreated, treated) - untreated["Y"].values

ate = np.mean(np.concatenate([effects_treated, effects_untreated]))
att = effects_treated.mean()
atu = effects_untreated.mean()
optimal = effects_untreated.max()

print(f"ATE (all items):          {ate:.4f}")
print(f"ATT (treated only):       {att:.4f}")
print(f"ATU (untreated only):     {atu:.4f}")
print(f"Optimal (max untreated):  {optimal:.4f}")
```
